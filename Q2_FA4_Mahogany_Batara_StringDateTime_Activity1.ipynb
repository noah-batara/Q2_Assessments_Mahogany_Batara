```python
# Activity 1: Student Details Formatter

# Input Data
first_name = input("Enter First Name: ")
last_name = input("Enter Last Name: ")
section = input("Enter Section: ")
fav_subject = input("Enter Favorite Subject: ")

# Cleaning Inputs
first_name = " ".join(first_name.split()).title()
last_name = " ".join(last_name.split()).title()
section = section.strip().upper()
fav_subject = " ".join(fav_subject.split()).title()

# Formatting Logic
full_name = f"{first_name} {last_name}"
full_name_upper = full_name.upper()
formatted_section = section.replace(" ", "")
formatted_subject = fav_subject.capitalize()
char_count = len(full_name.replace(" ", ""))

# Display Output
print("\n--- Formatted Output ---")
print(f"Name: {full_name}")
print(f"Name Code: {full_name_upper}")
print(f"Section: {formatted_section}")
print(f"Favorite Subject: {formatted_subject}")
print(f"Character Count: {char_count}")
```
