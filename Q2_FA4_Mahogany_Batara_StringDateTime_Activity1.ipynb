```python
# Activity 2: Deadline & Date Formatter

from datetime import datetime

# Input Data
event_title = input("Enter Event Title: ").strip().title()
year = int(input("Enter Year: "))
month = int(input("Enter Month: "))
day = int(input("Enter Day: "))

# Create datetime object
event_date = datetime(year, month, day)
today = datetime.now()

# Date Formatting
formatted_date = event_date.strftime("%B %d, %Y")
weekday = event_date.strftime("%A")
month_name = event_date.strftime("%B")

# Calculate Days Remaining
days_left = (event_date.date() - today.date()).days

# Determine Deadline Status
if days_left > 0:
    status = "Upcoming"
    elif days_left == 0:
        status = "Due Today!"
        else:
            status = "Past Due"

            # Display Output
            print("\n===== EVENT SCHEDULE CARD =====")
            print(f"Event Title: {event_title}")
            print(f"Date: {formatted_date}")
            print(f"Year: {year}")
            print(f"Month: {month_name}")
            print(f"Weekday: {weekday}")
            print(f"Days Remaining: {abs(days_left) if days_left < 0 else days_left}")
            print(f"Status: {status}")
            print("===============================")
```
